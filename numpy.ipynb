```python
# House Price Prediction

# house = [Area(sq.ft), Bedrooms, price]
house = np.array([
    [500,1,25],
    [800,2,40],
    [1000,2,50],
    [1200,3,65],
    [1500,3,82],
    [2000,4,95],
])

# Get area and bedrooms from house
hprice = house[:,:-1]

# Area Weight = [per sqft price, per bedroom price]
weights = np.array([0.05,5])

# Calculate Predictions
predictions = hprice @ weights

# print('House Price is ', predictions)

# Iterating array
for price in predictions:
  print('House Price is:', price)
```
